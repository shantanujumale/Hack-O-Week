```python
import pandas as pd

# Create DataFrame
data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Rahul", "Sneha", "Rohan", "Priya"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Math": [78, 88, 65, 92, 55, 80],
    "Science": [82, 91, 70, 94, 60, 85],
    "English": [75, 90, 68, 96, 58, 84],
    "Attendance": [92, 95, 80, 98, 70, 90],
    "StudyHours": [3, 5, 2, 6, 1, 4]
}

df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

```python
# ============================================
# STUDENT PERFORMANCE ANALYSIS PROJECT
# Technologies: Python, NumPy, Pandas,
#               Matplotlib, Seaborn
# ============================================



import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


# --------------------------------------------
# Step 2: Data Cleaning
# --------------------------------------------

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# Remove duplicate records
df.drop_duplicates(inplace=True)

print("\n========== DATA TYPES ==========")
print(df.dtypes)



# --------------------------------------------
# Step 4: NumPy Operations
# --------------------------------------------

marks = df[["Math", "Science", "English"]].to_numpy()

print("\n========== NUMPY ANALYSIS ==========")
print("Overall Mean :", np.mean(marks))
print("Maximum Mark :", np.max(marks))
print("Minimum Mark :", np.min(marks))
print("Standard Deviation :", np.std(marks))

# Vectorized Operation
updated_marks = marks * 1.05

print("\n========== MARKS AFTER 5% BONUS ==========")
print(updated_marks)

# --------------------------------------------
# Step 5: Function Example
# --------------------------------------------

def topper(data):
    return data.loc[data["Average"].idxmax()]


print("\n========== TOPPER ==========")
print(topper(df))

# --------------------------------------------
# Step 6: Object-Oriented Programming
# --------------------------------------------

class StudentAnalysis:

    def __init__(self, dataframe):
        self.df = dataframe

    def average_marks(self):
        return self.df["Average"].mean()

    def topper(self):
        return self.df.loc[self.df["Average"].idxmax()]

    def failed_students(self):
        return self.df[self.df["Average"] < 60]

    def highest_attendance(self):
        return self.df.loc[self.df["Attendance"].idxmax()]

    def subject_average(self):
        return self.df[["Math", "Science", "English"]].mean()


analysis = StudentAnalysis(df)

print("\n========== CLASS OUTPUT ==========")
print("Class Average:", analysis.average_marks())

print("\nTop Student")
print(analysis.topper())

print("\nFailed Students")
print(analysis.failed_students())

print("\nHighest Attendance")
print(analysis.highest_attendance())

print("\nSubject Average")
print(analysis.subject_average())

# --------------------------------------------
# Step 7: List Comprehension
# --------------------------------------------

df["Status"] = ["Pass" if x >= 40 else "Fail" for x in df["Math"]]

# --------------------------------------------
# Step 8: Dictionary Comprehension
# --------------------------------------------

subjects = ["Math", "Science", "English"]

subject_average = {subject: df[subject].mean() for subject in subjects}

print("\n========== SUBJECT AVERAGES ==========")
print(subject_average)

# --------------------------------------------
# Step 9: GroupBy
# --------------------------------------------

print("\n========== AVERAGE MARKS BY GENDER ==========")
print(df.groupby("Gender")["Average"].mean())

print("\n========== ATTENDANCE BY GENDER ==========")
print(df.groupby("Gender")["Attendance"].mean())

# --------------------------------------------
# Step 10: Sorting
# --------------------------------------------

print("\n========== TOP STUDENTS ==========")
top_students = df.sort_values(by="Average", ascending=False)

print(top_students)

# --------------------------------------------
# Step 11: Filtering
# --------------------------------------------

print("\n========== STUDENTS ABOVE 80 ==========")
print(df[df["Average"] > 80])

print("\n========== LOW ATTENDANCE (<75) ==========")
print(df[df["Attendance"] < 75])

# --------------------------------------------
# Step 12: Correlation Matrix
# --------------------------------------------

print("\n========== CORRELATION MATRIX ==========")
print(df.corr(numeric_only=True))

# --------------------------------------------
# Step 13: Summary Statistics
# --------------------------------------------

print("\n========== DESCRIPTIVE STATISTICS ==========")
print(df.describe())

# --------------------------------------------
# Step 14: Export Final Dataset
# --------------------------------------------

df.to_csv("student_performance_analysis.csv", index=False)

print("\nProcessed dataset saved as:")
print("student_performance_analysis.csv")

# --------------------------------------------
# Step 15: Data Visualization
# --------------------------------------------

# Set style
sns.set_style("whitegrid")

# --------------------------------------------
# Bar Chart - Average Marks
# --------------------------------------------

plt.figure(figsize=(8,5))
plt.bar(df["Name"], df["Average"])
plt.title("Average Marks of Students")
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.show()

# --------------------------------------------
# Subject Average
# --------------------------------------------

subject_avg = df[["Math", "Science", "English"]].mean()

plt.figure(figsize=(6,4))
subject_avg.plot(kind="bar")
plt.title("Subject-wise Average")
plt.ylabel("Marks")
plt.show()

# --------------------------------------------
# Attendance vs Average
# --------------------------------------------

plt.figure(figsize=(7,5))
plt.scatter(df["Attendance"], df["Average"])
plt.title("Attendance vs Average Marks")
plt.xlabel("Attendance")
plt.ylabel("Average Marks")
plt.show()

# --------------------------------------------
# Grade Distribution
# --------------------------------------------

plt.figure(figsize=(6,4))
sns.countplot(x="Grade", data=df)
plt.title("Grade Distribution")
plt.show()

# --------------------------------------------
# Correlation Heatmap
# --------------------------------------------

plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()

# --------------------------------------------
# Box Plot
# --------------------------------------------

plt.figure(figsize=(7,5))
sns.boxplot(data=df[["Math","Science","English"]])
plt.title("Marks Distribution")
plt.show()

# --------------------------------------------
# Histogram
# --------------------------------------------

plt.figure(figsize=(7,5))
plt.hist(df["Average"], bins=6)
plt.title("Average Marks Distribution")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()

# --------------------------------------------
# Pair Plot
# --------------------------------------------

sns.pairplot(
    df,
    vars=["Math", "Science", "English", "Attendance", "StudyHours"],
    hue="Gender"
)
plt.show()

# --------------------------------------------
# Final Report
# --------------------------------------------

print("\n================ FINAL REPORT ================")

print("Total Students :", len(df))
print("Overall Average :", round(df["Average"].mean(), 2))
print("Highest Average :", round(df["Average"].max(), 2))
print("Lowest Average :", round(df["Average"].min(), 2))

print("\nTopper")
print(df.loc[df["Average"].idxmax(), ["Name", "Average"]])

print("\nLowest Performer")
print(df.loc[df["Average"].idxmin(), ["Name", "Average"]])

print("\nGrade Counts")
print(df["Grade"].value_counts())

print("\nPass/Fail Counts")
print(df["Status"].value_counts())

print("\nProject Completed Successfully!")
```

```python
# --------------------------------------------
# Step 3: Create New Columns
# --------------------------------------------

df["Average"] = df[["Math", "Science", "English"]].mean(axis=1)


def grade(avg):
    if avg >= 90:
        return "A"
    elif avg >= 75:
        return "B"
    elif avg >= 60:
        return "C"
    else:
        return "D"


df["Grade"] = df["Average"].apply(grade)
```
